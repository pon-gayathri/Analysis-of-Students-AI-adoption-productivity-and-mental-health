```python
#correlation
# to check whether daily study hours and CGPA are related
#(ie.,does CGPA increases when study hours increases)

from scipy import stats

correlation, p_value = stats.pearsonr(students["Daily_Study_Hours"], students["CGPA"])

print(f"Correlation : {correlation}")
print(f"p-value : {p_value}")

#H0 : there is no relationship between study hours and CGPA(correlation = 0)
#H1 : there is a significant relationship between study hours and CGPA(correlation != 0)

if correlation > 0:
    print("There is a positive relationship")
elif correlation == 0:
    print("There is no relationship")
else:
    print("There is a negative relationship")

#p_value

alpha = 0.05

if p_value < alpha:
    print("Reject H0, there is a relationship between study hours and CGPA")
else:
    print("Failed to reject H0, there is no relationship between study hours and CGPA")
   
```

```python
#Two sample t-test - two tailed
# To check whether the CGPA of students with high ai dependency level and low ai dependency level are same or different.

High_ai_dependency = students[students["Ai_Dependency_Level"] == "High"]["CGPA"]
Low_ai_dependency = students[students["Ai_Dependency_Level"] == "Low"]["CGPA"]

print(High_ai_dependency)
print(Low_ai_dependency)

print(f"Average CGPA of students with High_ai_dependency = {High_ai_dependency.mean()}")
print(f"Average CGPA of students with Low_ai_dependency = {Low_ai_dependency.mean()}")

from scipy.stats import ttest_ind

#H0 : There is no difference between the average CGPA of students with high ai dependency level and low ai dependency level
#H1 : There is a significant difference between the average CGPA of students with high ai dependency level and low ai dependency level

t_statistic, p_value = ttest_ind(High_ai_dependency, Low_ai_dependency)
print(f"t-statistic = {t_statistic}") # average cgpa of students with high ai dependency level is very slightly
                                      # (only in numeric values but generally no difference) lower than 
                                      # that of students with low ai dependency.
print(f"p-value = {p_value}")

alpha = 0.05

if p_value < alpha:
    print("Reject H0, there is a significant difference  between the CGPA of students with high ai dependency level and low ai dependency level")
else:
    print("Failed to reject H0,There is no difference between the CGPA of students with high ai dependency level and low ai dependency level")
```

```python
#probability
# There are 3000 students.If i choose one student at random, what is probability of 
# choosing a student with productivity score more than 85?

total_students = len(students)
print(total_students)
productivity_score_greater_than_85 = len(students[students["Productivity_Score"] >= 85])
print(productivity_score_greater_than_85)

probability = productivity_score_greater_than_85/total_students
print(probability)
print(f"The probability of choosing a student at random with productivity score more than 85 is {round(probability * 100,2)}%")
```

```python
#Anova
# To check if there is a difference in atleast one among the average AI replaceability fear of students in different branches.
unique_branch = students["Branch"].unique()
print(unique_branch)

branchwise_ai_replaceability_fear = []
for ub in unique_branch:
    ai_fear = students[students["Branch"] == ub]["Ai_Replaceability_Fear"]
    print(ub)
    print(ai_fear)
    average = ai_fear.mean()
    print(f"Average : {average}")
    branchwise_ai_replaceability_fear.append(ai_fear)
print(branchwise_ai_replaceability_fear)

#H0 : There is no difference among the average AI replaceability fear of students in different branches 
#H1 : There is a difference in atleast one among the average AI replaceability fear of students in different branches

from scipy.stats import f_oneway

f_statistic, p_value = f_oneway(*branchwise_ai_replaceability_fear)
print(f"f_statistic : {f_statistic}")
print(f"p-value : {p_value}")

alpha = 0.05
if p_value < alpha:
    print("Reject H0, There is a difference in atleast one among the average AI replaceability fear of students in differnt branches")
else:
    print("Failed to reject H0, There is no difference among the average AI replaceability fear of students in differnt branches")
          
```
